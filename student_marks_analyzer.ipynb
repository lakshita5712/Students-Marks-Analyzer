```python
import numpy as np
import pandas as pd
```

```python
students=np.array(["Kashi","Lavi","Sid","Jack","Rocky","Siya","Ram","Reena","Koshik","Komal","Krati","Priya","Gauri","Kanha","Shiv"])
subjects="Python","CyberSecurity","DataScience","Cloudcomputing","Ai"
```

```python
marks=np.array([
    [61,78,89,90,89],
    [95,96,97,98,99],
    [46,37,28,38,45],
    [87,35,56,47,67],
    [27,35,37,44,67],
    [61,78,89,90,34],
    [95,76,87,38,33],
    [26,37,28,38,22],
    [87,35,56,47,78],
    [27,35,37,44,33],
    [61,78,89,90,67],
    [95,76,87,38,67],
    [26,37,28,38,32],
    [87,35,56,47,78],
    [27,35,37,44,45],
])
print(marks)
```

```python
#Pandas DataFrame for tabular representation of Data
df=pd.DataFrame(marks,index=(students),columns=(subjects))
dfs
```

```python
#Total Marks of students out of 500
total_marks=np.sum(marks,axis=1)
print(total_marks)
```

```python
# Student Ranking 
rank_indices = np.argsort(-total_marks)

print("----- Student Ranking -----")

for rank, index in enumerate(rank_indices, start=1):
    print(f"Rank {rank}: {students[index]} - Total Marks = {total_marks[index]}")
```

```python
# Average Marks of Studenta
average_marks=np.mean(marks,axis=1)
print(average_marks)
```

```python
#Percentage Analysis
percentage=(total_marks/500)*100
print(percentage)
highest_percentage_index=np.argmax(percentage)
highest_percentage=percentage[highest_percentage_index]
print(highest_percentage)
lowest_percentage_index=np.argmin(percentage)
lowest_percentage=percentage[lowest_percentage_index]
print(lowest_percentage)
```

```python
a=np.where(percentage>=40,"Pass","Fail")
print(a)
```

```python
#Highest Marks Calculation
highest_index=np.argmax(total_marks)
print("Rollno:",highest_index)
h_name=students[highest_index]
print("Name:",h_name)
print("Highest Marks:",total_marks[highest_index],"out of 500")
print("Percentage: ",highest_percentage)
result=a[highest_index]
print("result:",result)
```

```python
#Lowest Marks Calculation
lowest_index=np.argmin(total_marks)
print("Rollno:",lowest_index)
l_name=students[lowest_index]
print("Name:",l_name)
print("Lowest Marks:",total_marks[lowest_index],"out of 500")
print("Percentage:",lowest_percentage)
result=a[lowest_index]
print("result:",result)
```

```python
#Subjectwise Analysis
subject_average=np.mean(marks,axis=0)
print(subject_average)
```
